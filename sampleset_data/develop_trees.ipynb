```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
import matplotlib.ticker as mticker
from matplotlib.colors import ListedColormap, BoundaryNorm


def prepare_scatter_data(cbfs_maxes_per_graph_size, graph_sizes, num_runs):
    xs = [[g_size] * num_runs for g_size in graph_sizes]

    all_ys = np.unique(np.concatenate(cbfs_maxes_per_graph_size))
    all_x, all_y, all_counts = [], [], []

    for x_row, y_row in zip(xs, cbfs_maxes_per_graph_size):
        counts = Counter(y_row)
        for y_val, count in counts.items():
            all_x.append(x_row[0])
            all_y.append(y_val)
            all_counts.append(count)

    unique_counts = sorted(set(all_counts))
    color_indices = [unique_counts.index(c) for c in all_counts]

    cmap = ListedColormap(
        plt.get_cmap("plasma_r")(np.linspace(0, 1, len(unique_counts)))
    )
    norm = BoundaryNorm(
        boundaries=np.arange(len(unique_counts) + 1) - 0.5,
        ncolors=len(unique_counts)
    )

    return all_x, all_y, all_ys, color_indices, unique_counts, cmap, norm


# === Dane ===
cbfs_er = [graphs_results_er[g_size].cbfs_maxes for g_size in graph_sizes]
cbfs_pwl = [graphs_results_pwl[g_size].cbfs_maxes for g_size in graph_sizes]

data_er = prepare_scatter_data(cbfs_er, graph_sizes, num_runs)
data_pwl = prepare_scatter_data(cbfs_pwl, graph_sizes, num_runs)

# === Figura z 2 osiami ===
fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=False)

# ---------- ERDOS-RENYI ----------
ax = axes[1]
scatter_er = ax.scatter(
    data_er[0], data_er[1],
    c=data_er[3],
    cmap=data_er[5],
    norm=data_er[6],
    s=100
)

ax.set_title(
    "Erdos-Renyi graphs\n\n"
    "CBF max. values of each iterative run (20)\n"
    "in function of graph size"
)
ax.set_xticks(graph_sizes)
ax.set_yticks(data_er[2])
ax.set_xlabel("Graph sizes")
ax.set_ylabel("CBF maxes from 20 hierarchical runs")
ax.yaxis.set_major_locator(mticker.MaxNLocator(integer=True))

cbar_er = plt.colorbar(scatter_er, ax=ax, ticks=range(len(data_er[4])))
cbar_er.ax.set_yticklabels(data_er[4])
cbar_er.set_label("Number of occurrences")

ax = axes[0]
scatter_pwl = ax.scatter(
    data_pwl[0], data_pwl[1],
    c=data_pwl[3],
    cmap=data_pwl[5],
    norm=data_pwl[6],
    s=100
)

ax.set_title(
    "Powerlaw graphs\n\n"
    "CBF max. values of each iterative run (20)\n"
    "in function of graph size"
)
ax.set_xticks(graph_sizes)
ax.set_xlabel("Graph sizes")
ax.set_ylabel("CBF maxes from 20 hierarchical runs")
ax.yaxis.set_major_locator(mticker.MaxNLocator(integer=True))

cbar_pwl = plt.colorbar(scatter_pwl, ax=ax, ticks=range(len(data_pwl[4])))
cbar_pwl.ax.set_yticklabels(data_pwl[4])
cbar_pwl.set_label("Number of occurrences")

plt.tight_layout()
plt.show()
```
